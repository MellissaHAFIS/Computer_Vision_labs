# 3. Understand annotation formats

## 3.1 Pascal VOC format

The original dataset uses XML files. A bounding box is stored as:

```text
xmin, ymin, xmax, ymax
```

These are pixel coordinates.

## 3.2 YOLO format

YOLO expects one `.txt` file per image. Each row has:

```text
class_id x_center y_center width height
```

Important: YOLO coordinates are **normalized** between 0 and 1.

Example:

```text
1 0.52 0.43 0.20 0.30
```

This means:

- class ID = 1
- center x = 52% of image width
- center y = 43% of image height
- box width = 20% of image width
- box height = 30% of image height

# 4. Convert BCCD annotations to YOLO format

The following cells create the YOLO-compatible dataset structure:

```text
bccd_yolo/
├── images/
│   ├── train/
│   ├── val/
│   └── test/
├── labels/
│   ├── train/
│   ├── val/
│   └── test/
└── data.yaml
```


````python
CLASS_NAMES = ["RBC", "WBC", "Platelets"]
CLASS_TO_ID = {name: idx for idx, name in enumerate(CLASS_NAMES)}

def voc_box_to_yolo(box, img_w, img_h):
    """
    Convert a Pascal VOC box [xmin, ymin, xmax, ymax] in pixels
    to YOLO normalized format [x_center, y_center, width, height].
    """
    xmin, ymin, xmax, ymax = box

    x_center = ((xmin + xmax) / 2) / img_w
    y_center = ((ymin + ymax) / 2) / img_h
    width = (xmax - xmin) / img_w
    height = (ymax - ymin) / img_h

    return x_center, y_center, width, height

def parse_voc_xml(xml_path):
    """Parse one Pascal VOC XML annotation file."""
    tree = ET.parse(xml_path)
    root = tree.getroot()

    size = root.find("size")
    img_w = int(size.find("width").text)
    img_h = int(size.find("height").text)

    objects = []
    for obj in root.findall("object"):
        class_name = obj.find("name").text
        if class_name not in CLASS_TO_ID:
            continue

        bbox = obj.find("bndbox")
        xmin = int(float(bbox.find("xmin").text))
        ymin = int(float(bbox.find("ymin").text))
        xmax = int(float(bbox.find("xmax").text))
        ymax = int(float(bbox.find("ymax").text))

        objects.append({
            "class_name": class_name,
            "class_id": CLASS_TO_ID[class_name],
            "box_xyxy": [xmin, ymin, xmax, ymax],
            "box_yolo": voc_box_to_yolo([xmin, ymin, xmax, ymax], img_w, img_h)
        })

    return img_w, img_h, objects
````

````python
def prepare_yolo_dataset(max_images=160, train_ratio=0.70, val_ratio=0.20):
    # Remove previous converted dataset to keep the notebook reproducible 😉
    if YOLO_DIR.exists():
        shutil.rmtree(YOLO_DIR)

    for split in ["train", "val", "test"]:
        (YOLO_DIR / "images" / split).mkdir(parents=True, exist_ok=True)
        (YOLO_DIR / "labels" / split).mkdir(parents=True, exist_ok=True)

    image_paths = sorted((RAW_DIR / "JPEGImages").glob("*.jpg"))
    image_paths = image_paths[:max_images]
    random.shuffle(image_paths)

    n = len(image_paths)
    n_train = int(train_ratio * n)
    n_val = int(val_ratio * n)

    splits = {
        "train": image_paths[:n_train],
        "val": image_paths[n_train:n_train + n_val],
        "test": image_paths[n_train + n_val:]
    }

    split_counts = {}
    class_counter = Counter()

    for split, paths in splits.items():
        split_counts[split] = len(paths)

        for img_path in paths:
            xml_path = RAW_DIR / "Annotations" / f"{img_path.stem}.xml"
            img_w, img_h, objects = parse_voc_xml(xml_path)

            # Copy image
            dst_img = YOLO_DIR / "images" / split / img_path.name
            shutil.copy2(img_path, dst_img)

            # Write YOLO label
            label_path = YOLO_DIR / "labels" / split / f"{img_path.stem}.txt"
            with open(label_path, "w") as f:
                for obj in objects:
                    class_counter[obj["class_name"]] += 1
                    x_center, y_center, width, height = obj["box_yolo"]
                    f.write(
                        f"{obj['class_id']} "
                        f"{x_center:.6f} {y_center:.6f} "
                        f"{width:.6f} {height:.6f}\n"
                    )

    data_yaml = {
        "path": str(YOLO_DIR),
        "train": "images/train",
        "val": "images/val",
        "test": "images/test",
        "names": {i: name for i, name in enumerate(CLASS_NAMES)}
    }

    with open(YOLO_DIR / "data.yaml", "w") as f:
        yaml.safe_dump(data_yaml, f, sort_keys=False)

    return split_counts, class_counter

split_counts, class_counter = prepare_yolo_dataset(MAX_IMAGES)

print("YOLO dataset created at:", YOLO_DIR)
print("Split counts:", split_counts)
print("Object counts:", class_counter)

print("\nData YAML:")
print((YOLO_DIR / "data.yaml").read_text())
````

# 6. Visualize YOLO labels

The function below converts YOLO normalized boxes back to pixel coordinates and draws them.

````python
def yolo_to_xyxy(x_center, y_center, width, height, img_w, img_h):
    """
    Convert YOLO normalized box to pixel box [xmin, ymin, xmax, ymax].

    Steps:
      1. De-normalize x_center and y_center by multiplying by image dimensions.
      2. Compute half-extents from width/height.
      3. Subtract / add half-extents to get the corner coordinates.
      4. Clamp to valid pixel range [0, img_w] and [0, img_h].
    """
    xmin = (x_center - width  / 2) * img_w
    ymin = (y_center - height / 2) * img_h
    xmax = (x_center + width  / 2) * img_w
    ymax = (y_center + height / 2) * img_h

    # Clamp to image boundaries (guards against tiny floating-point overflows)
    xmin = max(0, int(xmin))
    ymin = max(0, int(ymin))
    xmax = min(img_w, int(xmax))
    ymax = min(img_h, int(ymax))

    return xmin, ymin, xmax, ymax
````

````python
def load_yolo_labels(label_path):
    labels = []
    if not Path(label_path).exists():
        return labels

    with open(label_path, "r") as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 5:
                continue
            class_id = int(parts[0])
            x_center, y_center, width, height = map(float, parts[1:])
            labels.append([class_id, x_center, y_center, width, height])
    return labels

def draw_yolo_labels(image_path, label_path):
    image = cv2.imread(str(image_path))
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    img_h, img_w = image.shape[:2]

    labels = load_yolo_labels(label_path)

    for class_id, x_center, y_center, width, height in labels:
        xmin, ymin, xmax, ymax = yolo_to_xyxy(x_center, y_center, width, height, img_w, img_h)
        xmin, ymin, xmax, ymax = map(int, [xmin, ymin, xmax, ymax])

        cv2.rectangle(image, (xmin, ymin), (xmax, ymax), (255, 0, 0), 2)
        cv2.putText(
            image,
            CLASS_NAMES[class_id],
            (xmin, max(20, ymin - 5)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (255, 0, 0),
            2
        )

    plt.figure(figsize=(7, 7))
    plt.imshow(image)
    plt.axis("off")
    plt.show()

# After completing yolo_to_xyxy, run this cell
sample_image = random.choice(list((YOLO_DIR / "images" / "train").glob("*.jpg")))
sample_label = YOLO_DIR / "labels" / "train" / f"{sample_image.stem}.txt"

print("Sample image:", sample_image.name)
draw_yolo_labels(sample_image, sample_label)
````

# 7.  IoU

IoU means **Intersection over Union**.

It measures how much a predicted box overlaps with a ground-truth box.

```text
IoU = area(intersection) / area(union)
```


# 8. Implement IoU

Complete the function below.

Input boxes use the pixel format:

```text
[xmin, ymin, xmax, ymax]
```

````python
def compute_iou(box_a, box_b):
    """
    Compute IoU between two boxes.
    box_a and box_b are in [xmin, ymin, xmax, ymax] format.
    """
    # Intersection coordinates: take the inner corners
    x_left   = max(box_a[0], box_b[0])
    y_top    = max(box_a[1], box_b[1])
    x_right  = min(box_a[2], box_b[2])
    y_bottom = min(box_a[3], box_b[3])

    # If boxes don't overlap, intersection is zero
    inter_w    = max(0, x_right - x_left)
    inter_h    = max(0, y_bottom - y_top)
    inter_area = inter_w * inter_h

    # Area of each individual box
    area_a = (box_a[2] - box_a[0]) * (box_a[3] - box_a[1])
    area_b = (box_b[2] - box_b[0]) * (box_b[3] - box_b[1])

    # Union = sum of areas minus the double-counted intersection
    union = area_a + area_b - inter_area

    # Guard against degenerate boxes (zero area)
    if union == 0:
        return 0.0

    return inter_area / union
````

**Does the pretrained model already detect RBC, WBC, or Platelets?**
No.

**Why or why not?**
The pretrained YOLOv8n weights were trained on **COCO**, a dataset of everyday photographs. Its learned features and output heads are tuned to recognise COCO objects, not microscopy content. When it sees a blood cell image, it has no "blood cell" class to predict, so it falls back to the closest visual pattern it knows — a cluster of round, blob-like shapes gets misclassified as `person` (or sometimes `sports ball`, `clock`, etc.) because those are the COCO categories whose feature activations happen to fire most strongly on that texture.

This is a classic example of a **domain gap**: the visual domain of microscope slides (uniform background, repetitive circular structures, no natural scene context) is completely different from the natural images COCO was built on.

**What classes was the pretrained model probably trained on?**
The 80 COCO classes, which include everyday objects such as:
- People, animals (dog, cat, bird…)
- Vehicles (car, bus, bicycle…)
- Household objects (chair, bottle, laptop…)
- Food items (banana, pizza…)

None of those are remotely related to blood cells, which is exactly why **fine-tuning on the BCCD dataset is necessary** — the model needs to adapt its final detection heads (and partially its backbone features) to this new visual domain and these three new class labels.

**1. What happens when the confidence threshold is very low?**
At conf=0.10, the model returns 26 RBCs, 1 WBC, and 2 Platelets. More boxes are

At conf=0.75, only 3 RBCs, 1 WBC, and 1 Platelet survive. The model becomes very

conf=0.25 or conf=0.50 appear to give the best balance on this image. At 0.25 we

boxes are likely tighter and cleaner. A threshold around 0.25–0.50 is generally
the recommended range for this dataset.

**4. How does the confidence threshold affect false positives?**
Lowering the threshold increases false positives: the model keeps boxes even when
it has low confidence, so uncertain or spurious detections are not filtered out.
Raising the threshold reduces false positives because only high-confidence
predictions survive.

**5. How does the confidence threshold affect false negatives?**
Raising the threshold increases false negatives: real objects whose predicted
confidence falls below the cutoff are silently dropped. This is especially
problematic for Platelets, which are small and rare — at conf=0.75 only 1 remains
detected versus 2 at conf=0.10, confirming the model is less confident about
Platelet detections overall. Lowering the threshold reduces false negatives at
the cost of more noise.

# 15. Compare predictions with ground truth using IoU

We will now compare model predictions to ground-truth boxes.

A prediction is usually considered correct if:

```text
class is correct AND IoU >= 0.5
```

This is a simplified evaluation, not full mAP.

````python
def get_ground_truth_boxes(image_path):
    label_path = YOLO_DIR / "labels" / "test" / f"{image_path.stem}.txt"
    labels = load_yolo_labels(label_path)

    image = cv2.imread(str(image_path))
    img_h, img_w = image.shape[:2]

    gt = []
    for class_id, x_center, y_center, width, height in labels:
        box = yolo_to_xyxy(x_center, y_center, width, height, img_w, img_h)
        gt.append({
            "class_id": class_id,
            "class_name": CLASS_NAMES[class_id],
            "box": list(map(float, box))
        })
    return gt

def get_prediction_boxes(image_path, conf=0.25):
    pred = model.predict(
        source=str(image_path),
        conf=conf,
        imgsz=IMG_SIZE,
        device=DEVICE,
        verbose=False
    )[0]

    detections = []
    if pred.boxes is None:
        return detections

    for box, cls, score in zip(pred.boxes.xyxy.cpu().numpy(),
                               pred.boxes.cls.cpu().numpy(),
                               pred.boxes.conf.cpu().numpy()):
        class_id = int(cls)
        detections.append({
            "class_id": class_id,
            "class_name": CLASS_NAMES[class_id] if class_id < len(CLASS_NAMES) else str(class_id),
            "confidence": float(score),
            "box": box.tolist()
        })
    return detections
````

````python
# TODO 5:
# Complete a simple matching procedure.
#
# For each prediction:
# 1. Compare it with every ground-truth box of the same class.
# 2. Compute IoU.
# 3. Keep the maximum IoU.
# 4. Mark it as correct if max IoU >= 0.5.

eval_image = random.choice(test_images)
gt_boxes = get_ground_truth_boxes(eval_image)
pred_boxes = get_prediction_boxes(eval_image, conf=0.25)

print("Image:", eval_image.name)
print("Ground truth boxes:", len(gt_boxes))
print("Predicted boxes:", len(pred_boxes))

for i, pred in enumerate(pred_boxes):
    best_iou = 0.0

    for gt in gt_boxes:
        if gt["class_id"] == pred["class_id"]:
            iou = compute_iou(pred["box"], gt["box"])
            if iou > best_iou:
                best_iou = iou

    status = "correct" if best_iou >= 0.5 else "incorrect"

    print(
        f"Prediction {i}: class={pred['class_name']}, "
        f"conf={pred['confidence']:.2f}, best IoU={best_iou:.3f}, status={status}"
    )
````

**Answers:**

**1. Can a high-confidence prediction still be incorrect?**
In this specific output, all high-confidence predictions happen to be correct —
predictions 0 to 4 (conf ≥ 0.50) all have IoU well above 0.5. However, this is
not always the case. A high confidence score only means the model is certain about
what it sees; it does not guarantee good localization. For example, if the model
confidently places a box slightly off-center on a densely packed cluster of RBCs,
the IoU with any individual ground-truth box could still fall below 0.5, making
it incorrect despite high confidence. Confidence and localization accuracy are two
independent quantities.

**2. Can a box be well localized but have the wrong class?**
Yes. IoU only measures spatial overlap — it says nothing about the predicted class.
A predicted box could perfectly overlap a ground-truth Platelet (IoU = 0.95) but
be labelled as RBC, in which case our matching procedure would not count it as
correct because we filter by class_id == gt["class_id"] before computing IoU.
In this output we cannot directly observe this since all WBC and RBC predictions
match their ground-truth class, but it is a real failure mode, especially between
RBC and Platelet which can appear visually similar at low confidence levels.

# 16. Non-Maximum Suppression intuition

Object detectors often predict multiple boxes around the same object.

**Non-Maximum Suppression**, or NMS, removes duplicate boxes.

Basic idea:

1. Keep the box with the highest confidence.
2. Remove other boxes that overlap too much with it.
3. Repeat.

The overlap is measured using IoU.

# 17. Implement a simple NMS



now you will implement a simplified version of NMS.

Object detectors may predict several boxes around the same object.  
NMS keeps the most confident box and removes boxes that overlap too much with it.

The algorithm is:

1. Sort boxes by confidence score.
2. Select the box with the highest score.
3. Remove all boxes with IoU greater than a threshold.
4. Repeat until no boxes remain.

````python
def simple_nms(detections, iou_threshold=0.5):
    # TODO 1: sort by confidence score, highest first
    detections = sorted(detections, key=lambda x: x["score"], reverse=True)

    kept = []

    while len(detections) > 0:
        # TODO 2 & 3: take the most confident box and keep it
        best = detections.pop(0)
        kept.append(best)

        remaining = []

        for det in detections:
            # TODO 4: different class → always keep
            # same class → keep only if IoU is below threshold
            if det["class_id"] != best["class_id"]:
                remaining.append(det)
            else:
                iou = compute_iou(best["box"], det["box"])
                if iou < iou_threshold:
                    remaining.append(det)

        detections = remaining

    return kept
````

````python
# Test it
example_detections = [
    {"box": [50, 50, 150, 150], "score": 0.95, "class_id": 0},
    {"box": [60, 60, 155, 155], "score": 0.80, "class_id": 0},
    {"box": [200, 200, 300, 300], "score": 0.90, "class_id": 0},
    {"box": [55, 55, 148, 148], "score": 0.70, "class_id": 0},
    {"box": [52, 52, 151, 151], "score": 0.88, "class_id": 1},
]

kept_boxes = simple_nms(example_detections, iou_threshold=0.5)

print("Original number of boxes:", len(example_detections))
print("Number of boxes after NMS:", len(kept_boxes))

for box in kept_boxes:
    print(box)
````
